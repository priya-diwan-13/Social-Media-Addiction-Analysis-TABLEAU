```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_excel("Uncleaned_social_media_dataset_with_userid.xlsx")
```

```python
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python

df['Sleep_Hours'] = pd.to_numeric(df['Sleep_Hours'], errors='coerce')
df['Productivity_Score'] = pd.to_numeric(df['Productivity_Score'], errors='coerce')


numeric_cols = df.select_dtypes(include=['int64','float64']).columns

df[numeric_cols] = df[numeric_cols].fillna(df[numeric_cols].mean())

print(df.head(50))
#print(df[numeric_cols])
```

```python
print(df[numeric_cols].isnull().sum())
```

```python

object_cols = df.select_dtypes(include='object').columns

object_cols = object_cols.drop('Date')

df[object_cols] = df[object_cols].fillna("Unknown")
df.head(50)

```

```python
df['Date'] = df['Date'].ffill()
df.head(50)
```

```python
print(df[object_cols].isnull().sum())
```
